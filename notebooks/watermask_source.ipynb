```python
import cartopy
import cartopy.crs as ccrs
import matplotlib.pyplot as plt
import cartopy.feature as cfeature
```

```python
ax = plt.subplot(projection=ccrs.PlateCarree())

coast = cfeature.GSHHSFeature(scale='full')
ax.add_feature(coast, facecolor="lightgray")

ax.add_feature(cartopy.feature.NaturalEarthFeature("physical", "land", "10m"),
              ec="red", fc="yellow", lw=2, alpha=0.4)

ax.add_feature(cartopy.feature.NaturalEarthFeature("physical", "ocean", "10m"),
              ec="w", fc="blue", lw=2, alpha=0.4, zorder=100)

coordinates = (49.2, 50.4, 39.6, 40.5)

ax.set_xlim(coordinates[0], coordinates[1])
ax.set_ylim(coordinates[2], coordinates[3])
ax.gridlines(crs=ccrs.PlateCarree(), linewidth=2, color='black', alpha=0.5, linestyle='--', draw_labels=True)
```
